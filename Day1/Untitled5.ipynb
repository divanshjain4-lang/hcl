```python
# unit=33
# price=3000
# target=12

# revenue=unit*price
# achievement=revenue/target
# print(achievement)
# print(f"revenue")
# print("")
# print(revenue)
# print("")
#def hi(n):
  #return
  # DAY 01 - SALES PERFORMANCE PROGRAM

# Function 1: Take inputs from the user
def get_inputs():

    while True:
        try:
            # Ask for units sold
            units_input = input("Enter units sold (or type 'end' to stop): ")

            if units_input.lower() == "end":
                return None

            units = int(units_input)

            # Ask for price per unit
            price_input = input("Enter price per unit (or type 'end' to stop): ")

            if price_input.lower() == "end":
                return None

            price = float(price_input)

            # Ask for sales target
            target_input = input("Enter sales target (or type 'end' to stop): ")

            if target_input.lower() == "end":
                return None

            target = float(target_input)

            # Return all three inputs
            return units, price, target

        except ValueError:
            print("Invalid input! Please enter valid numeric values.")
            print("Please enter all details again.\n")


# Function 2: Calculate sales performance
def sales_performance(units, price, target):

    try:
        # Calculate total revenue
        revenue = units * price

        # Calculate target achievement percentage
        achievement = (revenue / target) * 100

        # Display results
        print("\n----- SALES PERFORMANCE REPORT -----")
        print(f"Units Sold: {units}")
        print(f"Price Per Unit: ₹{price:.2f}")
        print(f"Sales Target: ₹{target:.2f}")
        print(f"Total Revenue: ₹{revenue:.2f}")
        print(f"Target Achievement: {achievement:.2f}%")

        if achievement >= 100:
            print("Status: Target Achieved!")
        else:
            print("Status: Target Not Achieved")

        print("------------------------------------\n")

    except ZeroDivisionError:
        print("Error: Sales target cannot be zero.")
        print("Please enter the details again.\n")


# Main program: Repeat until user types 'end'
while True:

    result = get_inputs()

    # Stop the program if user enters 'end'
    if result is None:
        print("Program ended. Thank you!")
        break

    # Pass inputs to sales_performance function
    units, price, target = result

    sales_performance(units, price, target)
```
